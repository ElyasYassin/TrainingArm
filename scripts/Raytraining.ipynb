```python
import gym
import myosuite
import ray
from ray.rllib.algorithms.ppo import PPOConfig
from ray.tune.registry import register_env

# 1. Define the environment creation function
def make_env(env_config=None):  # env_config required by Ray
    return gym.make("CenterReachOut-v0")

make_env()
```

```python
# 2. Register the environment with Ray
register_env("center_reach", make_env)

# 3. Initialize Ray (only needed if not already running)
ray.init(ignore_reinit_error=True)

# 4. Configure PPO
config = (
    PPOConfig()
    .environment(env="center_reach")  # must match registered name
    .framework("torch")               # or "tf2"
    .rollouts(num_rollout_workers=4) # use 0 to debug
    .resources(num_gpus=1)
    .training(model={"fcnet_hiddens": [512, 512]})
)

# 5. Build and train
algo = config.build()
results = algo.train()

print(f"Reward after first iteration: {results['episode_reward_mean']}")
```
